# Part 12.2: Few Shot and Chain of Thought

In this module, we explore two well-known prompting techniques: few-shot prompting and chain-of-thought prompting. Both come from the early years of large language models, and both still appear constantly in documentation, tutorials, and production prompts. The models themselves have changed a great deal since these techniques were introduced, so we will cover how each one works and also how much of it remains useful with a modern reasoning model such as GPT-5.6.

Few-shot prompting gives the model a small number of worked examples before asking it to handle a new case. The examples show the model the format you expect and, just as important, calibrate its judgment. In a life insurance context, a handful of example risk classifications shows an underwriting assistant where the boundaries between categories lie. Modern models follow written instructions far better than early models did, so tasks that once required examples can often be handled with a clear instruction alone. Examples still earn their place when the format is unusual or when the decision boundary is hard to describe in words.

Chain-of-thought prompting asks the model to work through a problem step by step before giving its answer. From roughly 2022 through 2024, this simple instruction measurably improved accuracy on math and logic problems, because it pushed models that would otherwise jump straight to an answer into spending computation on intermediate steps. Reasoning models changed this picture. A model like GPT-5.6 reasons internally before it writes anything, whether you ask it to or not. Adding "think step by step" to a prompt no longer improves its accuracy, and prescribing the steps yourself can even get in the way. What the technique still offers is transparency. You can ask the model to write its reasoning into the answer so that a human can review it. In insurance, where decisions must be explained to clients, auditors, and regulators, that written rationale is often the product you actually need.

### Few-Shot Prompting

In the context of life insurance, few-shot prompting is a practical way to show the model exactly what you want. The model is given a handful of examples (typically 2 to 5) that demonstrate the desired output before it processes a new input. The examples guide the model toward responses that match your format and your standards. For life insurance underwriters, few-shot prompting can streamline tasks such as analyzing applicant data, explaining policy options, or estimating risk factors.

For example, an underwriter might want the model to classify applicants into different risk categories based on their medical history. A few-shot prompt could look like this:

Prompt Example 1:
```
Classify the following applicants based on their medical history:

Applicant 1: Age 45, history of smoking, mild hypertension.
Classification: Moderate Risk.

Applicant 2: Age 30, non-smoker, no significant medical history.
Classification: Low Risk.

Applicant 3: Age 55, history of diabetes and heart disease.
Classification: High Risk.

Applicant 4: Age 50, history of high cholesterol.
Classification:
```

By including several examples of applicants with varying health profiles, the model is primed to understand the relationship between health data and risk categories. The AI can then apply these patterns to new cases, assisting underwriters in identifying risk factors efficiently.

In another case, a few-shot prompt can be used to help an AI generate policy explanations. An underwriter might provide examples of simple and clear policy summaries:

Prompt Example 2

```
Summarize these life insurance policies in a single sentence:

Policy A: $500,000 coverage, 30-year term, premiums increase after 10 years.
Summary: A 30-year term policy with $500,000 coverage, with premiums that increase after the first 10 years.
Policy B: $250,000 coverage, whole life policy with guaranteed premiums.
Summary: A whole life policy providing $250,000 coverage with guaranteed stable premiums.
Policy C: $350,000 coverage, CUL life policy.
Summary:
```

These examples help the model understand how to summarize policy details in a concise and understandable manner, useful for underwriters explaining complex products to clients.

A final use case might involve estimating life expectancy based on health factors, which is a crucial part of risk assessment for life insurance policies. A few-shot prompt might include examples like:

Prompt Example 3

```
Estimate the life expectancy for the following individuals based on their health profiles:

Person A: Age 60, non-smoker, exercises regularly, no significant medical conditions.
Life Expectancy: 85 years.
Person B: Age 45, smoker, history of high cholesterol and hypertension.
Life Expectancy: 70 years.
Person C: Age 35, heavy smoker, history of cancer and diabetes
Life Expectancy:
```

In this case, few-shot prompting provides the model with clear benchmarks for different health scenarios, helping underwriters estimate life expectancy more consistently. Each example in the prompt conditions the model to replicate accurate patterns in its future predictions.

One caution for modern applications. When the categories are fixed, as in the risk classification example above, the structured output techniques of Module 5 enforce the category list outright, since a Literal typed field cannot return a label you did not define. Few-shot examples and schemas work well together. The schema guarantees the shape of the answer, while the examples calibrate the judgment that fills it in.

### Chain of Thought Prompting

Chain-of-thought prompting asks the model to articulate its reasoning step by step as part of its answer. When the technique was introduced, it made models measurably more accurate on multi-step problems. With a modern reasoning model, the accuracy benefit has largely disappeared, because the model already reasons internally before producing its first visible word. What the instruction controls today is the output. It determines whether the reasoning gets written down where a person can read it.

That is still valuable, and in regulated fields it can be essential. When an underwriter asks the model to assess a potential client's risk based on age, health history, lifestyle, and family medical history, requesting the reasoning produces a documented rationale for the file. This ensures each factor is addressed in writing and gives a reviewer something concrete to check.

```
Prompt:
Assess the risk level for a 52-year-old male applicant who has a history of hypertension, is a non-smoker, exercises regularly, and has a family history of heart disease. Please explain your reasoning step by step.

Model Response:
Let's break down the risk assessment step by step:

Age: The applicant is 52 years old, which places him in the middle-aged category. While age alone does not immediately indicate high risk, it is a factor that needs to be considered, as health risks tend to increase with age.

Health history: The applicant has a history of hypertension. This condition is a significant risk factor for heart disease, stroke, and other cardiovascular problems. However, the severity of the hypertension and how well it is managed (e.g., through medication or lifestyle changes) would further influence this risk.

Lifestyle: The applicant is a non-smoker, which significantly reduces the risk of heart disease, lung disease, and cancer. Additionally, the fact that he exercises regularly is a positive factor, as regular physical activity helps control blood pressure, manage weight, and improve overall cardiovascular health. This healthy lifestyle choice can partially mitigate the risks posed by his hypertension.

Family medical history: The applicant has a family history of heart disease. This is an important consideration because it suggests a genetic predisposition to cardiovascular issues. Even though the applicant's own lifestyle choices are healthy, this hereditary factor could increase his long-term risk.

Final assessment: Taking all factors into account, the applicant is not in the highest risk category due to his non-smoking status and regular exercise. However, his history of hypertension and family predisposition to heart disease moderately increase his risk. Therefore, I would classify him as moderate risk, with the recommendation to further monitor his blood pressure management and cardiovascular health over time.
```

Notice the phrasing in the prompt. It asks the model to explain its reasoning, which is a request about what appears in the answer. That is the right way to use this technique with a reasoning model. The older phrasing, "think step by step," gives instructions about a process the model now performs on its own, and it adds nothing on a model like GPT-5.6.

Three practical points complete the picture. First, a written rationale is the model's account of its conclusion, and a human should still review it before it supports a real underwriting decision. Second, the small local models you met in Module 8 generally lack built-in reasoning, and classic chain-of-thought prompting still improves their accuracy the way it improved frontier models in 2023. Third, reasoning models often expose a setting that controls how much internal reasoning they perform. Adjusting that setting, rather than the prompt, is the modern way to trade cost and speed against depth on hard problems.

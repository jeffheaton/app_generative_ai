# Part 11.1: Introduction to the Model Context Protocol

In Module 7 you built agents and gave them tools: a web search, a calculator, a button that drives a toy car. Every one of those tools was wired directly into your own program. That works, but consider what happens at industry scale. Suppose there are N applications that can host an agent (ChatGPT, Claude, Cursor, your Streamlit app) and M useful capabilities to connect (GitHub, a database, a file system, your company's inventory system). Wiring each capability into each host requires N x M custom integrations, each with its own bugs, its own maintenance, and its own security review. Every team rebuilds the same GitHub connector, slightly differently, forever.

The **Model Context Protocol (MCP)** is the industry's answer to this problem. Introduced by Anthropic in late 2024 and since adopted across the ecosystem -- OpenAI, Google, Microsoft, and essentially every major agent framework now support it -- MCP is an open standard that turns N x M into N + M. A capability is packaged once, as an **MCP server**; any application that speaks the protocol can use it. The common analogy is USB-C: one standardized plug, and any device works with any charger. You will often see MCP described as "the USB-C port for AI applications."

## The Architecture

MCP has three roles:

* **Host** - The application the user actually interacts with: a chat app, an IDE, or the agents you build in this course.
* **Client** - The protocol machinery inside the host that maintains a connection to one server. A host runs one client per server.
* **Server** - A program that exposes capabilities. Servers are usually small and focused: one wraps a file system, another wraps GitHub, another wraps your database.

The host asks each server, over a standard handshake, "what can you do?" -- and the server answers with a machine-readable catalog. Nothing about the host needs to change when a server adds a capability, and nothing about the server needs to know which host is calling it.

## The Three Primitives

An MCP server can expose three kinds of things:

| Primitive | What it is | Who decides to use it |
|---|---|---|
| **Tools** | Functions the model can call, with typed parameters | The model, during reasoning |
| **Resources** | Data identified by a URI that can be read into context | The application (or user) |
| **Prompts** | Reusable prompt templates the server provides | The user, typically via UI |

Tools will feel immediately familiar: they are exactly the tools you hand-built in Module 7, and their parameter schemas are exactly the Pydantic-style schemas you designed in Module 5 -- MCP simply moves them behind a protocol boundary. Resources are the quieter, equally important half: a way for servers to offer *data* (a file, a report, a database row) without the model having to call a tool to get it.

## Transports

A client and server talk over one of two transports:

* **stdio** - The client launches the server as a subprocess and speaks to it over standard input/output. Zero network setup; ideal for local capabilities and for the classroom.
* **Streamable HTTP** - The server runs anywhere and the client connects over HTTP, with standard web authentication. This is how hosted and third-party servers are consumed.

We will use stdio throughout this module because it is fully self-contained inside Colab: our notebooks will write a server to disk, and the agent's client will launch it automatically.

## The Ecosystem in 2026

MCP's growth since 2024 has been remarkable. Thousands of servers are published for everything from Postgres to Blender; official registries make them discoverable; and on the Python side, the FastMCP framework -- which we will use in Part 11.3 -- powers the large majority of servers in the wild. When a vendor ships a new product integration for AI today, it ships as an MCP server. This is why MCP belongs in a course on *applications* of generative AI: it is the packaging format in which real-world capabilities now arrive.
